# ChatScene P0 Query Human Review Workbench

这个 notebook 用于图形化完成 300 个 query subject 的 P0 人工审计：先完成 48 个 development query，再完成 252 个 test query。

边界：

- 无需 API key、GPU、CARLA 或网络。
- 界面按五步审核：参与者与外部事件 → 道路与空间 → 规则与风险 → 最终一致性确认 → CPD 与备注。
- 宽屏下请求原文在左侧保持可见，右侧完成五步审核；窄屏自动改为单栏。
- 每张要求卡直接展示自然语言评分要求、依据和严格程度，常用判断不再需要先展开卡片；内部 atom 字段默认隐藏。
- `source` 是机器候选，`target` 只在人工修改、拆分或合并时表示修订后的最终要求。
- 常用判断是单击“正确”“不应计分”和“允许但不强制”；“其他判断”中保留修改、拆分、合并及撤回选择，复杂修订仍使用结构化字段。
- 要求卡显示当前选择，各步骤标明待选数量；“已选/已选齐”只表示选择进度，不代表校验通过或人工确认。
- 顶部“标注指南”集中说明计分严格程度、判断依据和完成规则；查看指南不会改变答案。
- 可从顶部“同内容批审”入口展开面板，处理同内容 requirement 和同 surface 的同策略 CPD；每一批会完整显示覆盖的所有 query 原文，批量操作只填未决定草稿，已有判断自动保留为实例例外。
- 普通接受会自动生成标准审计依据；四项 atom 一致性总结由正式 before/after projection 自动派生，并在完成 subject 时整体确认。support/response 仍需人工选择，源数据修改或错误始终要求手写理由。
- CPD 修订先选择结构化原因，只有“其他原因”或拒绝 CPD source 时才显示自由文本。
- 点击“完成本条审核”或“完成并继续”失败时，会在当前 query 上方列出每张问题卡、具体缺口和定位按钮，并自动定位第一张问题卡；必要时展开其高级编辑区。
- 第4步不再要求重复选择四项 atom 总结；requirements 改变时对应总结自动变为 revise，未改变时自动为 accept。若自动投影或 ego gate/surface style 等非 atom 源字段有误，可勾选异常并填写理由，subject 会保持阻塞。CPD 总结继续与第5步的唯一 CPD 判断同步。
- 每张 requirement 卡会列出它自动关联的第4步总结；第4步每项也可展开查看具体关联卡。关联可以重叠，这是正式一致性关联规则，不是重复审核，也不是要求你手工给评分指标分配字段。
- 第5步的 CPD 结论与 policy 不一致时，清单会直接说明是“声称修改但未实际变化”还是“源数据错误导致阻塞”，定位到 CPD，并且不显示内部 check 名。
- 同一 intent 先完成 precise；通过正式逐题校验后，partial/vague 只会继承尚未填写的审核槽，已有实例判断和已完成项绝不会被覆盖。继承后仍保持未完成，必须按各自文本修改差异并显式完成。
- “完成并继续”通过正式校验后，按当前筛选和搜索范围跳到下一条未完成请求；末尾会回绕，没有其他待处理项时提示，校验或保存失败时留在当前条目。
- Agent semantic draft 和机器预填都只是参考，不是 human gold。
- 保存草稿只会写入 source-bound checkpoint；每个 subject 必须由人显式校验并标记完成。
- 只有整个 split 全部完成、输入 `FINALIZE` 并通过 canonical finalizer 后，才会生成一份/subject 的 human gold。
- Reviewer ID 是任务分配标签；真实身份与产物保管由外部人工流程保证。
- 同一 assignment 只打开一个浏览器标签页 / kernel；关闭页面前先点“保存草稿”。
